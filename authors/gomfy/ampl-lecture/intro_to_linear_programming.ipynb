```python
!pip install -q amplpy

from amplpy import AMPL

ampl = AMPL()

ampl.eval("""
var X1 >= 0;
var X2 >= 0;
var X3 >= 0;
var X4 >= 0;

maximize Z:
0.115*X1 + 0.065*X2 + 0.220*X3 + 0.070*X4;

subject to Presupuesto:
X1 + X2 + X3 + X4 <= 50;

subject to ASFI:
X1 + X2 + X4 >= 25;

subject to Consumo:
X3 <= 7.5;

subject to Microcredito:
X1 >= 20;
""")

print("Modelo cargado correctamente")
```

```python
ampl.getOutput("option solver highs;")
print(ampl.getOutput("solve;"))
print(ampl.getOutput("display X1, X2, X3, X4, Z;"))
```
